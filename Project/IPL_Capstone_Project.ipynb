# Loading the Libraries and Dataset

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('IPL.csv')
df.head()
```

1. Which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y=match_wins.index, x=match_wins.values, palette='viridis')
plt.title("Most match win by team")
```

2. Toss Decision Trends

```python
# df["toss_decision"]
# sns.barplot(x=df['toss_decision'], palette='viridis')  # Wrong
# sns.histplot(x = df['toss_decision'],palette='rainbow', data=df)
sns.countplot(x = df['toss_decision'],palette='rainbow', data=df)
plt.title('Toss Decision Trends')
```

3. Toss Winner vs Match Winner

```python
win = df[df['toss_winner']==df['match_winner']]['match_id'].count()
percentage = (win * 100)/df.shape[0]
percentage.round(2)
```

4. How do teams win? (Runs vs Wickets)

```python
sns.countplot(x='won_by',data=df)
plt.title("Won by")
```

4. 10 Best Bowling Figures

```python
df.head(0)
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=top_bowlers.values, y=top_bowlers.index, palette="mako")
# top_bowlers.plot(kind = 'barh')
```
