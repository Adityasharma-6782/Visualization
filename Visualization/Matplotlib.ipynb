**Matplotlib Basic**

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)    # line graph
```

```python
plt.plot(x,y,)
plt.title("Hello")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

**Subplot**

```python
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4)   # only 1 to 4 due to (2,2)
```

```python
plt.subplot(2,2,1)
plt.plot(x==y, y==x)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4) 
plt.plot(x==2, y==2)
```

```python
plt.subplot(6,6,1)
plt.subplot(6,6,2)
plt.subplot(6,6,3)
plt.subplot(6,6,5)
plt.subplot(6,6,4)
plt.subplot(6,6,6)
plt.subplot(6,6,10)
```

**Object Oriented Way**

==> Matplotlib alllow the aspect ratio, DPI and figure size to be specified when Figure object is created. You can use the figure and dpi keyword argument.

==> figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (Pixels-per-inch).

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])  # (left, bottom, width, height)
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.5,0.3,0.3])
axis2.plot(x,y)
axis3 = fig.add_axes([0.6,0.2,0.2,0.2])
axis3.plot(x,y)
```

```python
# fig.add_axes([0.1,0.1,0.2,0.8])
```

```python
plt.figure().add_axes([0.1,0.1,0.8,0.8]).plot(x,y)
```
