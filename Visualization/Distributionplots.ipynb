```python
import matplotlib.pyplot as plt
import seaborn as sns
```

We use distribution plots to visualize the distribtion of quantitative data.

Let's discuss some plots that allow us to vidualise the distribtion of a data set. These plots are:

==>    *displot/histplot*

==>    *joinplot*

==>    *pairplot*

==>    *rugplot*\

```python
df =sns.load_dataset('tips')
df
```

```python
df['size'].unique()    # it also consider as qualitative 
```

```python
# sns.distplot(df['total_bill'])
# sns.distplot(df['total_bill'],bins=20)      # bins ==> is use for total rectangle
sns.histplot(df['total_bill'],bins=20,kde=True)       # kde= False ==> It is use for linr on the hist
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20,kde=True)
# sns.histplot(df,bins=20,kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde=True)
```

**jont plot**

for combining two or more plots

```python
# sns.jointplot(['total_bill', 'tip'])  # It is wrong
# sns.jointplot(x='total_bill', y='tip', data=df)
sns.jointplot(x='total_bill', y='tip', data=df,kind='hex',color= 'black')  # kind ==> by default 'scatter' --> { "scatter" | "kde" | "hist" | "hex" | "reg" | "resid" }
```

```python
sns.jointplot(x='total_bill', y='tip', data=df,kind='kde',color= 'black')
```

**Pair Plot**

use for comparison 

```python
# sns.pairplot(df, hue='sex',palette='rainbow')
sns.pairplot(df, hue='size',palette='rainbow')
```

**Rug Plot**

```python
# sns.rugplot(df['total_bill'])
sns.rugplot(df,palette='rainbow')
```
